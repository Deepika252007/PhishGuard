# PhishGuard: Phishing URL Risk Analyzer

**Goal:** demonstrate a small defensive cybersecurity tool that inspects URL text for suspicious patterns.

This is a rule-based heuristic, not a trained machine-learning model. The program never opens or fetches the submitted URL. A score is not proof of safety or maliciousness.

## 1. How it works

1. Normalize the URL by trimming whitespace and adding `https://` when no scheme is provided.
2. Parse the text using Python's standard-library URL parser.
3. Check for explainable warning patterns: IP-address host, `@` in the authority, excessive length, many subdomains, security-sensitive terms, known URL shorteners, and HTTP.
4. Add rule points and cap the score at 100.
5. Label scores below 25 as Low, 25–49 as Medium, and 50 or above as High.

These thresholds are simple teaching choices, not validated clinical/security probabilities.

```python
from scanner import analyze_url

examples = [
    "https://example.com",
    "https://example.org/account/verify",
    "http://192.0.2.1/login",
    "http://example.com@192.0.2.1/login",
]

for url in examples:
    result = analyze_url(url)
    print(f"URL: {url}")
    print(f"Risk: {result['risk']} ({result['score']}/100)")
    print("Reasons:")
    for reason in result["reasons"]:
        print(" -", reason)
    print()
```

## 2. Interpret the results

The example domains `example.com` and `example.org` are reserved for documentation. The IP address `192.0.2.1` is also a documentation-range address. They are useful for demonstrations without testing real suspicious sites.

A Low score means only that few configured rules matched. It does **not** certify a site as safe.

## Limitations and ethics

- Heuristics may produce false positives and false negatives.
- Legitimate websites can contain words such as “login” or “verify”.
- The application analyzes text only and does not connect to submitted addresses.
- Use this project for education and defensive awareness.
